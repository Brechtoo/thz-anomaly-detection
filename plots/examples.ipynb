```python
import pandas as pd
import matplotlib.pyplot as plt
from pathlib import Path


CSV_PATH = "/results/measurements/combined_dataset.csv"

df = pd.read_csv(CSV_PATH)

# Nur Label 1
subset = df[df["label"] == 1].sample(5, random_state=42)

for i, row in enumerate(subset.itertuples(), start=1):
    file_path = Path(row.file_path)

    raw = pd.read_csv(
        file_path,
        sep=r"\s+",
        header=None,
        comment="#",
        names=["x", "y"]
    )

    plt.figure(figsize=(10, 5))
    plt.plot(raw["x"], raw["y"])

    plt.xlabel("x")
    plt.ylabel("y")
    plt.title(f"Label 1 - Beispiel {i}", fontsize=10)
    plt.grid(True)
    plt.tight_layout()
    plt.show()
```
